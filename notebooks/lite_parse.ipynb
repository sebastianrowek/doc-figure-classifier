### Only Page Complexity

```python
parser = LiteParse(ocr_enabled=False)
pages = parser.is_complex("../data/reports/dgroup_ann_rep_2023.pdf")
```

```python
target_pages = []

for index, page in enumerate(pages):
    if page.layout.figure_count > 0 or page.layout.ruled_table_count > 0:
        target_pages.append(index+1)
        figure_count = page.layout.figure_count
        table_count = page.layout.ruled_table_count
        print(f"Page {index + 1}: {figure_count} figures, {table_count} tables")

target_pages_str = ",".join(str(page) for page in target_pages)
```

### Extract Page Layout
The `extract_blocks` argument for the LiteParse class is a feature of the 2.13.0 release.<br>
This realease is currently not available for my platform so we can't try it for now.

```python
parser = LiteParse(
    ocr_enabled=True,              # Enable OCR (default: True)
    ocr_language="de",            # Tesseract language code
    #max_pages=10,
    target_pages=target_pages_str,        # List of page numbers to process (default: all)
    output_format="json",          # "json" | "text" | "markdown"
    image_mode="placeholder",      # Markdown image handling: "placeholder" | "off" | "embed"
    extract_images=True,           # Extract image bytes + metadata (default: False)
    image_output_dir="../data/reports/images",   # Write images and return name/path metadata (optional)
    extract_annotations=False,     # Include page annotations in structured output
    extract_blocks=True,
)
```

```python
with open("../data/reports/dgroup_ann_rep_2023.pdf", "rb") as f:
    result = parser.parse(f.read())
```
